```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("student_data.csv")
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(x='Study_Hours', y='Marks', data=df, palette='viridis')

plt.title('Subject-wise Marks')
plt.xlabel('Study_Hours')
plt.ylabel('Marks Obtained')

plt.show()
```

```python

plt.figure(figsize=(8, 5))
sns.histplot(df['Marks'], bins=10, kde=True, color='blue')

# Labels mattrum Title
plt.title('Distribution of Student Marks')
plt.xlabel('Marks Range')
plt.ylabel('Number of Students')

# Plot-ai kaata
plt.show()
```

```python
# Scatter Plot uruvaaka
plt.figure(figsize=(8, 5))
sns.scatterplot(x='Study_Hours', y='Marks', data=df, color='red', marker='o')

# Labels mattrum Title
plt.title('Study Hours vs Student Marks')
plt.xlabel('Hours Studied')
plt.ylabel('Marks Obtained')

# Plot-ai kaata
plt.show()
```

```python
plt.figure(figsize=(6, 5))
# Sleep Hours-in spread-ai paarkka
sns.boxplot(y='Sleep_Hours', data=df, color='lightblue')

plt.title('Box Plot of Sleep Hours')
plt.ylabel('Hours of Sleep')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
# Maanavan peyar thavira matra numeric columns-ku correlation paarkkirom
numeric_df = df[['Study_Hours', 'Attendance_Percent', 'Screen_Time_Hours', 'Sleep_Hours', 'Marks']]
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt=".2f")

plt.title('Correlation Heatmap of Student Features')
plt.show()
```
